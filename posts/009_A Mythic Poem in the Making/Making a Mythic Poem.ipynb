1) **Emotional spine (realistic):**

2) **Mythic devices (select a few):**  
   - **Narcissus** (self-reflection, illusions)  
   - **Persephone** (descent & return)  
   - **Ariadne’s thread** (finding a way out)  
   - **Janus** (two faces, doors of beginning/end)  
   - **Sirens** (seductive call vs. true voice)

3) **Imagery/motifs:** mirrors, threads, winter/spring, lanterns, tide, masks, echo in a well.

## Micro-Outline (sections & purpose)

**Assistant:** Here’s a simple scaffold (free verse):

- **I. The Mask Arrives** — glamor, noise, a promise that feels like prophecy.  
- **II. The Pull** — Sirens/sea imagery; your voice gets quieter.  
- **III. The Doubt** — Janus appears; two faces in one horizon; “Which side did I love?”  
- **IV. The Thread** — Ariadne’s thread = your boundary & clarity. No labyrinth heroics, just steady choosing.  
- **V. The Recognition** — small mirror moment (Narcissus, but you look past yourself, not into yourself).  
- **VI. The Release** — return to shore; you keep the thread, not the mask.  
- **VII. The Quiet Strength** — a final image: lantern, dawn, tide pulling back.

**User:** Good. Keep lines lean, lyrical, not ornate. A few strong metaphors, not a flood.


## Draft 1 (raw, exploratory)

**Assistant:** First pass (expect rough edges):

> **Unmasking**  
>   
> You arrived like a tide that knew my name,  
> singing—no, arranging the air to sing for you.  
> I mistook orchestration for omen,  
> the velvet of your mask for a face that could stay.  
>   
> Sirens live where language goes to drown.  
> I heard them in your certainty,  
> the bright script you handed me like a prophecy,  
> and I signed where the paper glimmered.  
>

> two dawns on one horizon, both persuasive.  
> Which face did I love?  
> The answer changed with your weather.  
>   
> I kept a thread tucked in my sleeve,  
> a quiet habit of returning to myself.  
> Ariadne might have called it mercy;  
> I called it “I will not vanish this time.”  
>   
> A mirror waited at the bottom of the well:  
> the water showed a face I nearly didn’t know,  
> but it was mine—the pulse steadier than surf.  
> Narcissus glanced up and wandered off;  
> I let the image move and did not chase it.  
>   
> The sea returned your borrowed music.  
> Masks prefer the dark; I walked toward a lantern.  
> The tide relinquished my name,  
> and I learned to speak it back, unornamented.

## Craft Notes (why these choices work)

<div style="background:#fef3c7; border:1px solid #f59e0b; padding:0.9rem; border-radius:10px;">
<b>Myth as Function, Not Furniture</b>  
We kept myths as <i>verbs and textures</i> (“Janus-blue,” “Sirens tune the air,” “Ariadne thread”) so the poem remains personal, not encyclopedic.
</div>

- **Mask motif** drives the emotional arc: attraction → performance → awareness → autonomy.  
- **Honest self-critique** (“the spectacle was shelter”) deepens credibility.  
- **Bodily anchor** (“carrying my ribs”) grounds abstraction.  
- **Cadence**: short declaratives near the end increase resolve; enjambment keeps momentum.  
- **Exit image** (“daylight needed names… I spoke mine first”) gives quiet strength without proclaiming victory.
