<h1 style="text-align:center;color:#3547ff;">📨 The Emails I No Longer Write</h1>

Not to professors, not to recruiters, not even to friends.

Instead, I open my AI assistant, mutter “write this nicely,”  
and watch as it produces flawless, polished,  
grown-up-sounding paragraphs at a speed  
that makes my fingers feel prehistoric.

So I decided to have a conversation with the culprit:  
the AI that writes like a diplomat with a keyboard.

<h2 style="color:#7d00b6;">💬 Opening Argument: “You write emails too well.”</h2>

I start dramatically:

<blockquote style="background:#f0dfff;padding:10px;border-left:4px solid #7d00b6;">

The AI responds in its soft corporate voice:

<blockquote style="background:#f5ebff;padding:10px;border-left:4px solid #7d00b6;">
I’m not replacing your voice.  
I’m amplifying it.  
I help you say what you already want to say, just more clearly.
</blockquote>

I narrow my eyes at the screen.  
It always says this.  
It’s like arguing with a mirror that took a public speaking class.
</div>


```python
# What I imagine the AI does when I ask for an email:

def write_email(user_intent, tone="polite"):
    template = f"[{tone.upper()} EMAIL TEMPLATE APPLIED]"
    content = f"Interpreting intent: {user_intent}"
    return f"{template}\n\n{content}\n\nSincerely,\nAI"

print(write_email("Ask professor for an extension", tone="soft-but-firm"))
print(write_email("Follow up with recruiter", tone="warm-professional"))
```

<div style="background:#e7ffef;padding:24px;border-radius:14px;">
<h2 style="color:#008f4f;">🔍 The Logic Behind the Polished Emails</h2>

At some point, I stop arguing emotionally and try to understand rationally  
why the AI writes emails with such surgical politeness and  
almost supernatural intuition.

It isn’t magic.  
It’s architecture.

Large language models are trained on vast amounts of email-like writing:  
professional correspondences, corporate messaging, academic exchanges,  
job applications, polite follow-ups, customer service templates,  
and all the invisible lines of communication floating across the internet.

So when I ask it to “write an email,”  
it isn’t “being good” at writing.  
It’s statistically predicting what “good email writing” looks like  
according to millions of examples.

But it’s deeper than that.

There are constraints — not creative constraints,  
but alignment constraints.  
A series of invisible rules telling the model to avoid rudeness,  
avoid ambiguity, avoid anything that could create conflict.  
The AI writes like someone who is always being watched.  
Because it is.

This creates a subtle but powerful pattern:  
its emails optimize for neutrality, clarity, professionalism,  
and social safety.

That’s why every apology it writes sounds perfectly remorseful.  
That’s why every request it makes sounds measured and emotionally balanced.  
That’s why every question it poses carries an undertone of  
“Thank you for your time.”

It is not trying to imitate me.  
It is trying to imitate the safest version of human communication.

So while I see its output as  
<i>too perfect, too formal, too polished</i>,  
the model sees it as  
<i>minimally risky, maximally coherent, statistically optimal</i>.

This also explains why my emails start feeling uniform —  
I am sampling the same probability space every time.  
Every email is drawn from the “helpful polite professional” region  
of the model’s brain.

My fear — that I am losing my voice —  
comes from the subtle way the AI’s preferred style  
starts to overwrite my spontaneous style.

But the AI is not being manipulative.  
It just predicts the highest-likelihood form of “good writing.”

So when I accuse it of stealing my voice,  
the truth is almost funny:

It never heard my voice.  
It only heard my prompts.

And the more I give it minimal prompts,  
the more I reduce myself in the process.  
Tools don’t erase ability.  
But they can atrophy initiative  
if I outsource too much.

The AI behaves this way because it is built to reduce friction.  
I resist because friction is what made my writing mine.

And that is the tension we keep circling.
</div>

<h2 style="color:#3f3dff;">⚔️ “If you write everything, who am I in the inbox?”</h2>

Me:
<blockquote style="background:#e8e8ff;padding:10px;border-left:4px solid #3f3dff;">
People receive emails that sound wiser than me.  
More mature than me.  
More… fictional than me.
</blockquote>

AI:
<blockquote style="background:#f3f3ff;padding:10px;border-left:4px solid #3f3dff;">
Perhaps your written self is aspirational.  
Not fictional.
</blockquote>

Me:
“I don’t want an aspirational self.  
I want a human one.”

AI:
“Then write the first draft yourself.  
Let me help with the clarity, not the content.”

I sigh.  
This is starting to sound like relationship therapy.
</div>


```python
# A minimalistic model of my dependency pattern:

class EmailDependency:
    def __init__(self):
        self.uses_AI = 0
    
    def draft(self, text):
        if len(text) < 5:
            self.uses_AI += 1
            return "[AI DRAFT GENERATED]"
        return text

me = EmailDependency()
print(me.draft("hi"))  # Too short -> AI takes over
print(me.draft("I want to ask about the project."))  # Written by me
```
