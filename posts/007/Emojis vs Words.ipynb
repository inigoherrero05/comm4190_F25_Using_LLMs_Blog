## 🧩 Scene 4 — Layered Meaning with Emojis

Emojis aren’t just cute add-ons. They work like *visual idioms*.  
- Literal layer: 🍎 = apple.  
- Figurative layer: 🍎 = health, education, temptation.  
- Cultural layer: 🍎 = New York, “the Big Apple.”  

LLMs trained on tons of data can catch common ones — but the subtler the layer, the more likely they’ll fall flat.  

> Example:  
> *“He went full 🥒 during the meeting.”*  
> - US slang: chill.  
> - Elsewhere: confused.

## 🔍 Scene 8 — Why Do Emojis Trip LLMs?

Three big reasons:  

1. **Data Bias**: Training corpora underrepresent emojis compared to text.  
2. **Ambiguity**: One 🍑 = fruit, another 🍑 = peachy behind.  
3. **Context Loss**: Emojis rely on *tone* and *shared culture*.  

So models either:  
- Over-literalize (🥦 = broccoli, period).  
- Over-dramatize (🔥 = passion, hellfire, apocalypse).
